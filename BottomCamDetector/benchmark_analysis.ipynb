```python
from glob import glob
import numpy as np

# open benchmark logs folder
logs = glob("benchmark_logs/*")

inf_times = {}

for log in logs:
    with open(log, "r") as file:
        log_inf_times = []
        for line in file:
            split_string = line.split('-')
            if len(split_string) > 1:
                if split_string[0] == '[OBSDET] Bottom processing times ':
                    log_inf_times.append(float(split_string[1].split(',')[1].split(':')[1][:-3]))
        inf_times[log] = np.asarray(log_inf_times)
        
for log in inf_times:
    print(f"model {log.split('/')[-1]} has average inference speed of {np.mean(inf_times[log])} ms")
```
